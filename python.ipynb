### Logical Functions (Operators)

In Python, logical operators (`and`, `or`, `not`) are used to combine conditional statements or to reverse the boolean result of a condition. They are fundamental for controlling program flow based on multiple conditions.

*   **`and`**: Returns `True` if both statements are `True`.
*   **`or`**: Returns `True` if at least one of the statements is `True`.
*   **`not`**: Reverses the result; returns `False` if the result is `True`, and vice-versa.

````python
# Example of Logical Operators

x = 5
y = 10
z = 15

# Using 'and'
print(f"x < y and y < z: {x < y and y < z}") # True and True -> True
print(f"x < y and y > z: {x < y and y > z}") # True and False -> False

# Using 'or'
print(f"x > y or y < z: {x > y or y < z}")   # False or True -> True
print(f"x > y or y > z: {x > y or y > z}")   # False or False -> False

# Using 'not'
print(f"not (x < y): {not (x < y)}")          # not True -> False
print(f"not (x > y): {not (x > y)}")          # not False -> True
````

### Membership Functions (Operators)

Membership operators in Python are used to test whether a value or variable is found in a sequence (like strings, lists, tuples) or a collection (like sets, dictionaries).

*   **`in`**: Returns `True` if the specified value is found in the sequence/collection.
*   **`not in`**: Returns `True` if the specified value is NOT found in the sequence/collection.

````python
# Example of Membership Operators

my_list = [1, 2, 3, 4, 5]
my_string = "hello world"
my_dict = {'a': 1, 'b': 2}

# Using 'in'
print(f"3 in my_list: {3 in my_list}")             # True
print(f"6 in my_list: {6 in my_list}")             # False
print(f"'hello' in my_string: {'hello' in my_string}") # True
print(f"'x' in my_string: {'x' in my_string}")       # False
print(f"'a' in my_dict: {'a' in my_dict}")           # True (checks keys)
print(f"1 in my_dict.values(): {1 in my_dict.values()}") # True (checks values)

# Using 'not in'
print(f"6 not in my_list: {6 not in my_list}")     # True
print(f"3 not in my_list: {3 not in my_list}")     # False
print(f"'xyz' not in my_string: {'xyz' not in my_string}") # True
````

````python
# Electricity Bill Calculator

unit = int(input("Enter the number of units consumed: "))

if unit <= 50:
    bill = 0
elif unit <= 100:
    # For units 51 to 100, charge 2 rs/unit for units above 50
    bill = (unit - 50) * 2
elif unit <= 200:
    # For units 101 to 200, charge 5 rs/unit for units above 100
    # First 50 units are free, next 50 units (51-100) at 2 rs/unit
    bill = (50 * 2) + (unit - 100) * 5
else:
    # For units above 200, charge 10 rs/unit for units above 200
    # First 50 units free, next 50 units (51-100) at 2 rs/unit, next 100 units (101-200) at 5 rs/unit
    bill = (50 * 2) + (100 * 5) + (unit - 200) * 10

print("Electricity Bill = ₹", bill)
````

````python
# outer join-> right , left and  full outer join
import pandas as pd
df1=pd.DataFrame([[1,'amal','developer',55000],[2,'sanoj','marketing',40000],[3,'rahul','developer',45000],[4,'manjoj','developer',52000],[5,'akshay','marketing',35000],[6,'anu','developer',45000]],columns=['id','name','dept','salary'])
print(df1)
df2=pd.DataFrame([[3,'kochi'],[5,'banglore'],[11,'kochi'],[7,'tvm'],[2,'mumbai'],[6,'chennai'],[8,'kochi']],columns=['id','place'])
print(df2)
````

````python
# inner join
out=pd.merge(df1,df2,on='id',how='inner')
print(out)
````

````python
df3=pd.DataFrame([[3,'kochi'],[5,'banglore'],[11,'kochi'],[7,'tvm'],[2,'mumbai'],[6,'chennai'],[8,'kochi']],columns=['emp_id','place'])
print(df3)
````

````python
# LEFT OUTER JOIN
out=pd.merge(df1,df3,left_on='id',right_on='emp_id',how='left')
print(out)
````

````python

#inner join
out=pd.merge(df1,df3,left_on='id',right_on='emp_id',how='inner')
print(out)
````

````python
# RIGHT JOIN
out=pd.merge(df1,df3,left_on='id',right_on='emp_id',how='right')
print(out)
````

````python
# Full outer join
out=pd.merge(df1,df3,left_on='id',right_on='emp_id',how='outer')
print(out)
````

````python
class bankaccount:

    def __init__(self,account_no,balance):

        self.account_no=account_no

        self.__balance=balance

    def print_balance(self):

        print(self.__balance)

    def deposit(self,amount):

        if self.__validate_amount(amount):

            self.__update_balance(amount)

        else:

            print('invalid amount')

    def __validate_amount(self,amount):

        return amount>0

    def __update_balance(self,amount):

        self.__balance += amount

user=bankaccount(103,10000)

# print(user.__balance)

user.print_balance()

user.deposit(10000)

user.print_balance()

# user.__update_balance(10000)




````

````python
# @classmethod

A class method is a method that works with the class itself, rather than a particular object.

It receives the class as its first argument, usually named cls.




````

````python
class Student:

    school = "ABC School"

    @classmethod

    def show_school(cls):

        print(cls.school)



Student.show_school()

s1=Student()

s1.show_school()




````

````python
# class method can be used as an alternative constructor.(init also a constructor).
````

````python
# class method can be used as an alternative constructor

class Student:

    def __init__(self, name, age):

        self.name = name

        self.age = age

    @classmethod

    def from_string(cls, data):

        name, age = data.split(",")

        return cls(name, int(age))

'anu,25'

'arjun,20'

# s1=Student('anu,25')

s1=Student.from_string('anu,25')

print(s1.name)

print(s1.age)




````

````python
# @static method

A static method is a method that does not need the object (self) or class (cls).

It behaves almost like a normal function, but logically it belongs to the class.




````

````python
class Calculator:



    @staticmethod

    def add(a, b):

        return a + b





result = Calculator.add(10, 20)



print(result)




````

````python
# @property (acess a method like attribute.part of or like Encapsulation)
````

````python
class Student:

    def __init__(self, name):

        self._name = name

    def get_name(self):

        return self._name



s1 = Student('naveen')

print(s1.get_name())

print(s1._name)




````

````python
class Student:

    def __init__(self, name):

        self._name = name

    @property

    def get_name(self):

        return self._name



s1 = Student('naveen')

print(s1.get_name)




````

````python

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    pass
d=Dog()
d.sound()
c=Cat()
c.sound()
````
