```python
HYPOTHESIS TESTING
-----------------
used to make decision from the data.
problem: is the model better than old model ?

step 1: defining hypothesis

null hyposthesis(H0): no difference exists.
new model performance =old model performance
alternate hypothesis(H1): difference exists
new model performance > old model performance
new model performance < old model performance

step 2: choose significance level (alpha)

normally alpha =0.05 (5% risk for wrong decision)

step 3: calcualte test statistics.

step 4: calculate p value
p value tells that how likely is the result if  H0 were true.

p value< alpha--> reject H0.












```
